1. Read the file using spark dataframe reader API
1. Add Metadata Columns 
   - Source File
    - Ingestion Timestamp
1. Write to bronze delta table

```python
#Define source_file and table_name
source_file = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

```python
# Define the schema
constructors_schema = """constructorId STRING,
                         name STRING,
                         nationality STRING,
                         url STRING
                         """
```

```python
#read data from contructors file
constructors_df = (
    spark.read
        .format('json')
        .schema(constructors_schema)
        .option('mode','FAILFAST')
        .load(source_file)
)
```

```python
display(constructors_df)
```

```python
constructors_final_df= add_ingestion_metadata(constructors_df)
```

## Step 2 - Add Metadata Columns
- Source File
- Ingestion Timestamp

```python
(
    constructors_final_df
        .write
        .format('delta')
        .mode('overwrite')
        .saveAsTable(table_name)
)
```
