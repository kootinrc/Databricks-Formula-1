## Set-up the project environment for Formula1 Project
1. Create External Location databricks-course-ext-dl1-formula1
1. Create Catalog formula1
1. Create Schemas landing, bronze, silver and gold
1. Create Volume Files in the landing schema


```python
%fs ls 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/landing'
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databrickscourseextstg32_ext_dl1_formula1
URL 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'External location for the formula1 container';
```

```python
%sql
SHOW CATALOGS;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS formula1
   MANAGED LOCATION 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/' 
   COMMENT 'This is the main catalog for the formula1 project' ;
```

## Create Schema

```python
%sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/gold';  
```

```python
%sql 
USE CATALOG formula1;
select current_catalog();
SHOW SCHEMAS;
```

## Create a volume

```python
%sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databrickscourseextstg32.dfs.core.windows.net/landing';
```
